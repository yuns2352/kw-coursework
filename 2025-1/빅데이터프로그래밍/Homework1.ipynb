```python
# 빅데이터 프로그래밍 Homework #1
# 프랙탈 도형 출력 프로그램
# 이름 : 윤성근
# 학번 : (학번 생략)

def main():
    while(True):
        print("도형 만들기")
        print("1.도형 그리기")
        print("2.종료")
        
        choice = input("숫자를 선택하세요!")
        if choice == "1":
            draw_fractal()
        elif choice == "2":
            print("종료할게요!")
            break
        else:
            print("오류입니다! 다시 입력하세요.")

def draw_fractal():
    while(True):
        try:
            n = int(input("프랙탈 반복횟수를 입력하세요!(1~4): "))
            if 1 <= n <= 4:
                print_fractal(n)
                return 
            else:
                print("1과 4 사이의 숫자를 입력하세요.")
        except ValueError:
            print("숫자를 입력하세요.")

def print_fractal(n):
    size_of_fractal = 3 ** n

    틀 = []
    for i in range(size_of_fractal):
        행 = []
        for j in range(size_of_fractal):
            행.append(' ')
        틀.append(행)

    def single_fractal(x, y):
        모양 = [
            ' * ',
            '***',
            ' * '
        ]
        for i in range(3):
            for j in range(3):
                틀[x + i][y + j] = 모양[i][j]

    def multi_fractal(x, y, level):
        if level == 1:
            single_fractal(x, y)
        else:
            unit = 3 ** (level - 1)
            multi_fractal(x + unit, y, level - 1)
            multi_fractal(x, y + unit, level - 1)
            multi_fractal(x + unit, y + unit, level - 1)
            multi_fractal(x + unit, y + unit * 2, level - 1)
            multi_fractal(x + unit * 2, y + unit, level - 1)

    multi_fractal(0, 0, n)

    for 행 in 틀:
        print(*행, sep = '')

main()
            
            
            
        
        
        
        
        
            
            



                
            
    

    
    





```
