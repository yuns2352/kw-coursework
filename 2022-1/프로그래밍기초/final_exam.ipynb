# 구조체를 사용한 성적 관리 구현

입력파일 input.txt 명세
- 첫 줄은 메타정보이다
  - 입력 받을 줄 수
  - 과목 개수
- 한 줄에 적혀있는 정보
  - 이름: name (이름의 길이는 20 캐릭터 미만이다)
  - 학년: sch_yr
  - 과목1: class1
  - 과목2: class2
  - 과목3: class3
  - 과목4: class4

다음과 같이 출력되도록 프로그램을 작성하시오.
<pre>
name  sch_yr cls1 cls2 cls3 cls4   sum   avg
kim     1      70   75   72   71   288   72.00   
lee     4      60   99   30   45   234   58.50   
kang    3      95   60   85   75   315   78.75   
joo     2      90   80   95  100   365   91.25   
nam     2      85   75   87   82   329   82.25   

the highest avg = 91.25 (joo)
  </pre>

```python
%%writefile input.txt
5 4
kim   1  70 75 72 71
lee   4  60 99 30 45
kang  3  95 60 85 75
joo   2  90 80 95 100
nam   2  85 75 87 82
```

```python
%%writefile test.cpp
#include <stdio.h>
#include <stdlib.h>
#include <string.h>

int size;
int n_classes;

typedef struct
{
  char name[20];
  int sch_yr; 
  int classes[4];
  int sum;
  float avg;
} Student;

int addStudent(Student s[], int idx, char name[], int sch_yr, int classes[])
{
    if (idx >= size) {
        printf("Error : out of size\n");
        return 0;
    }

    strcpy(s[idx].name, name);
    s[idx].sch_yr = sch_yr;
    int sum = 0;
    for (int i = 0; i < n_classes; ++i) {
        s[idx].classes[i] = classes[i];
        sum += classes[i];
    }
    s[idx].sum = sum;
    float avg = 0;
    for (int i = 0; i < n_classes; ++i) {
        s[idx].classes[i] = classes[i];
        (sum += classes[i]) / n_classes;
    }
    s[idx].avg = avg;
}

int printStudent(Student s[])
{
     printf("name  sch_yr cls1 cls2 cls3 cls4   sum   avg\n");
         for (int i = 0; i < size; ++i) {
             printf("%-4s ", s[i].name);
             for (int k = 0; k < size; ++k) {
                 printf(" %-3d", s[i].sch_yr);
             }
             for (int j = 0; j < n_classes; ++j) {
             printf(" %-4d", s[i].classes[j]);
         }
         printf("%-3d", s[i].sum);
         printf("%-3f", s[i].avg);
         printf("\n");
     }
  printf("\n");
}

int getMaxAvg(Student s[])
{
    float max_avg = 0;
    int idx = -1;
    for (int i = 0; i < size; ++i) {
        max_avg += s[i].classes[idx];
    }
    return idx;
}

int main()
{
    FILE *fp = fopen("input.txt", "r");
    if (fp == NULL) exit(1);
    
    fscanf(fp, "%d %d", &size, &n_classes);
 
    Student *s = new Student[size];
 
    char name[20];
    int sch_yr, classes[4];
 
    for (int i = 0; i < size; ++i) {
        fscanf(fp, "%s", name);
        for (int j = 0; j < n_classes; ++j) {
            fscanf(fp, "%d", &classes[j]);
        }
        addStudent(s, i, name, sch_yr, classes);
    }
 
    printStudent(s);
 
    int max_idx = getMaxAvg(s);
    printf("the highest avg = %5.2f (%s)\n", s[max_idx].avg, s[max_idx].name);

    delete[] s;
    fclose(fp);
    return 0;
}
```
