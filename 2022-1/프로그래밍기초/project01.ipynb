# 숫자 카운트 다운

## Step 1

**입력처리**

input.txt에 적혀있는 정수들을 입력받는다.  
각 정수 x의 크기는 0 <= x <= 9 사이의 값으로 제한한다.
<pre>
5
9
0
</pre>

**출력처리**
입력받은 x를 한 줄을 띄워 출력한다.
<pre>
5

9

0
</pre>

입력파일

```python
%%writefile input.txt
5
9
0
```

다음 코드셀에 test.cpp 내용 작성

```python
%%writefile test.cpp
#include <stdio.h>
#include <stdlib.h>

int main()
{
    FILE *fp = fopen("input.txt", "r");
 
    if (fp == NULL) {
        printf("ERROR : file open failure.");
        exit(1);
    }
 
    int n[9];
    fscanf(fp, "%d", &n[0]);
    fscanf(fp, "%d", &n[1]);
    fscanf(fp, "%d", &n[2]);
 
    printf("%d\n\n", n[0]);
    printf("%d\n\n", n[1]);
    printf("%d\n\n", n[2]);
 
    fclose(fp);
 
    return 0;
}
```

코드 컴파일 및 실행

```python
!g++ test.cpp; ./a.out
```

## Step 2

**입력처리**

input.txt에 적혀있는 정수들을 입력받는다.  
각 정수 x의 크기는 0 <= x <= 9 사이의 값으로 제한한다.
<pre>
5
9
0
</pre>

**출력처리**

입력받은 x가 0이 될때까지 값을 변경하며 칸을 띄워 출력한다.  
각 x는 한 줄을 띄워 출력한다.
<pre>
5 4 3 2 1 0

9 8 7 6 5 4 3 2 1 0

0
</pre>

입력파일

```python
%%writefile input.txt
5
9
0
```

다음 코드셀에 test.cpp 내용 작성

```python
%%writefile test.cpp
#include <stdio.h>
#include <stdlib.h>

int main()
{
    FILE *fp = fopen("input.txt", "r");
 
    if (fp == NULL) {
        printf("ERROR : file open failure.");
        exit(1);
    }
 
    int n[9];
    fscanf(fp, "%d", &n[0]);
    fscanf(fp, "%d", &n[1]);
    fscanf(fp, "%d", &n[2]);
 
    printf("%d ", n[0]);
    while (n[0] > 0) {
        printf("%d ", --n[0]);
        if (n[0] == 0) {
            printf("\n\n");
        }
    } 
    printf("%d ", n[1]);
    while (n[1] > 0) {
        printf("%d ", --n[1]);
        if (n[1] == 0) {
            printf("\n\n");
        }
    } 
 
    printf("%d", n[2]);

    fclose(fp);
 
    return 0;
}
```

코드 컴파일 및 실행

```python
!g++ test.cpp; ./a.out
```

## Step 3

**입력처리**

input.txt에 적혀있는 정수들을 입력받는다.  
각 정수 x의 크기는 **-9 <= x <= 9** 사이의 값으로 제한한다.
<pre>
5
9
-7
0
-6
</pre>

**출력처리**

입력받은 x가 0이 될때까지 값을 변경하며 칸을 띄워 출력한다. (오른쪽 정렬)  
각 x는 한 줄을 띄워 출력한다.
<pre>
 5  4  3  2  1  0 

 9  8  7  6  5  4  3  2  1  0 

-7 -6 -5 -4 -3 -2 -1  0 

 0 

-6 -5 -4 -3 -2 -1  0 
</pre>

입력파일

```python
%%writefile input.txt
5
9
-7
0
-6
```

다음 코드셀에 test.cpp 내용 작성

```python
%%writefile test.cpp
#include <stdio.h>
#include <stdlib.h>

int main()
{
    FILE *fp = fopen("input.txt", "r");
 
    if (fp == NULL) {
        printf("ERROR : file open failure.");
        exit(1);
    }
 
    int n[9];
    fscanf(fp, "%d", &n[0]);
    fscanf(fp, "%d", &n[1]);
    fscanf(fp, "%d", &n[2]);
    fscanf(fp, "%d", &n[3]);
    fscanf(fp, "%d", &n[4]);
 
    printf("%d ", n[0]);
    while (n[0] > 0) {
        printf("%d ", --n[0]);
        if (n[0] == 0) {
            printf("\n\n");
        }
    } 
    printf("%d ", n[1]);
    while (n[1] > 0) {
        printf("%d ", --n[1]);
        if (n[1] == 0) {
            printf("\n\n");
        }
    } 
 
    printf("%d ", n[2]);
    while (n[2] < 0) {
        printf("%d ", ++n[2]);
        if (n[2] == 0) {
            printf("\n\n");
        }
    }

    printf("%d\n\n", n[3]);

    printf("%d ", n[4]);
    while (n[4] < 0) {
        printf("%d ", ++n[4]);
        if (n[4] == 0) {
            printf("\n\n");
        }
    }    

    fclose(fp);
 
    return 0;
}
```

코드 컴파일 및 실행

```python
!g++ test.cpp; ./a.out
```
