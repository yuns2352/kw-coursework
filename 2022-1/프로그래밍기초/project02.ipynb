# 01 - 평균 구하기

9주차 실습 코드 중 getSum() 함수를 참조하여  
평균을 구하는 프로그램을 작성하시오.


```python
%%writefile test.cpp
#include <stdio.h>

float getAverage(int *arr, int size) 
{
    float mean = 0;
    for (float i = 0; i < size; ++i) {
        mean += (*arr);
        ++arr;
    } 
    return mean; 
}

int main()
{
    int a[5] = {1, 2, 3, 4, 5};
 
    printf("average = %3.2f", getAverage(a, 5) / 5);
 
    return 0;
}
```

```python
!g++ test.cpp; ./a.out
```

실행 후 출력 결과는 다음과 같아야 한다
<pre>
average = 3.00
</pre>

# 02 - 문자열 복사

프로그램이 수행되도록 /*****/ 안을 채우시오.

```python
%%writefile test.cpp
#include <stdio.h>
#include <string.h>

void copyString(char source[], char target[])
{
    int i;
    for (i = 0; source[i] != 0; ++i) {
        target[i] = source[i];
    }
 
    target[i] = '\0';
}

int main()
{    
    char a[20] = "hello world";
    char b[20];
 
    copyString(a, b);
    
    printf("string: %s\n", &b[0]);
    printf("size  : %d\n", int(strlen(b)) );
    
    return 0;
}
```

```python
!g++ test.cpp; ./a.out
```

실행 후 출력 결과는 다음과 같아야 한다
<pre>
string: hello world
size  : 11
</pre>

첫 줄은 입력파일의 line 개수 정보가 있다.

입력파일

```python
%%writefile input.txt
5
1 6
2 7
3 8
4 9
5 10
```

프로그램이 수행되도록 /*****/ 안을 채우시오.

```python
%%writefile test.cpp
#include <stdio.h>
#include <stdlib.h>

int main()
{
    FILE *fp = fopen("input.txt", "r");
    if (fp == NULL) {
        printf("ERROR : file open failure.");
        exit(1);
    }        
    
    int size;
    fscanf(fp, "%d", &size);
 
    int a[5], b[5];
    int i = 0;
    while (i < size) {
        fscanf(fp, "%d %d", &a[i], &b[i]);
        ++i;
    }
    
    i = 0;
    while (i < size) {
        printf("%2d %2d\n", a[i], b[i]);
        ++i;
    }
 
    fclose(fp);
    return 0;
}
```

```python
!g++ test.cpp; ./a.out
```

실행 후 출력 결과는 다음과 같아야 한다
<pre>

2  7
3  8
4  9
5 10
</pre>

## 동적 배열 사용

프로그램이 수행되도록 /*****/ 안을 채우시오.
- *a는 malloc 을 사용
- *b는 new 를 사용

```python
%%writefile input.txt
5
1 6
2 7
3 8
4 9
5 10
```

```python
%%writefile test.cpp
#include <stdio.h>
#include <stdlib.h>

int main()
{
    FILE *fp = fopen("input.txt", "r");
    if (fp == NULL) exit(1);        
    
    int size;
    fscanf(fp, "%d", &size);
 
    
    int *a = (int *)malloc(sizeof(int) * size); 
    int *b = new int; 
 
    int i = 0;
    while (i < size) {
        fscanf(fp, "%d %d", &a[i], &b[i]);
        ++i;
    }
    
    i = 0;
    while (i < size) {
        printf("%2d %2d\n", a[i], b[i]);
        ++i;
    }
 
    free(a);
    delete[] b;
    fclose(fp);
    return 0;
}
```

```python
!g++ test.cpp; ./a.out
```

실행 후 출력 결과는 다음과 같아야 한다
<pre>
1  6
2  7
3  8
4  9
5 10
</pre>
